```python
import pandas as pd

train_df = pd.read_csv(
    "standardized_extra/std_favorita_large_train.csv",
    parse_dates=["ds"]
)

train_df.head()
```

```python
holidays_df = pd.read_csv(
    "favorita_covariates/holidays_events.csv",
    parse_dates=["date"]
)

oil_df = pd.read_csv(
    "favorita_covariates/oil.csv",
    parse_dates=["date"]
)

items_df = pd.read_csv(
    "favorita_covariates/items.csv"
)

stores_df = pd.read_csv(
    "favorita_covariates/stores.csv"
)
```

```python
oil_df = oil_df.rename(
    columns={
        "date": "ds"
    }
)

oil_df["dcoilwtico"] = (
    oil_df["dcoilwtico"]
    .ffill()
    .bfill()
)
```

```python
holidays_df = holidays_df.rename(
    columns={
        "date": "ds"
    }
)

holidays_df["is_holiday"] = 1

holiday_features = holidays_df[
    ["ds", "is_holiday"]
].drop_duplicates()
```

```python
df = train_df.merge(
    oil_df[["ds", "dcoilwtico"]],
    on="ds",
    how="left"
)

df = df.merge(
    holiday_features,
    on="ds",
    how="left"
)

df["is_holiday"] = (
    df["is_holiday"]
    .fillna(0)
    .astype(int)
)
```

```python
df["dayofweek"] = df["ds"].dt.dayofweek
df["month"] = df["ds"].dt.month
df["quarter"] = df["ds"].dt.quarter
df["year"] = df["ds"].dt.year
df["day"] = df["ds"].dt.day
```

```python
df = df.sort_values(
    ["unique_id", "ds"]
)

for lag in [1, 7, 14, 28]:

    df[f"lag_{lag}"] = (
        df.groupby("unique_id")["y"]
        .shift(lag)
    )
```

```python
for window in [7, 14, 28]:

    df[f"rolling_mean_{window}"] = (
        df.groupby("unique_id")["y"]
        .transform(
            lambda x:
            x.shift(1)
             .rolling(window)
             .mean()
        )
    )
```
